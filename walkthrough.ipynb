This notebook demonstrates the end-to-end autonomous agent workflow:
1. Loading input datasets (JSON & CSV)
2. Filtering & transforming data via Python comprehensions
3. Calling external REST APIs & Mock APIs
4. Generating structured analytical reports

```python
from agent import AutonomousAnalysisAgent

# 1. Initialize agent
agent = AutonomousAnalysisAgent(config_path='data/sample_input.json', force_mock=False)
tasks = agent.load_input_data()
print('Agent loaded successfully.')
print(f'Number of tasks: {len(tasks)}')
print(f'Categories: {agent.get_categories()}')
```

## Lesson 2: Data Comprehensions in Action
Demonstrating list comprehensions for priority filtering:

```python
high_priority_tasks = agent.filter_by_priority('HIGH')
print('High priority tasks:')
for t in high_priority_tasks:
    print(f' - [{t.task_id}] {t.target_topic}')
```

## Lesson 6: API Calls & Mock Fallbacks
Executing agent intelligence queries across external endpoints:

```python
results = agent.execute_tasks()
print('Execution results:')
for r in results:
    src = r.api_interaction.get('source', 'N/A')
    status = r.api_interaction.get('status', 'N/A')
    print(f' - {r.task_id}: {src} (Status: {status})')
```

## Lesson 7: Saving Outputs
Writing structured JSON and executive Markdown reports to `output/`:

```python
paths = agent.save_outputs('output')
print('Reports saved:')
print(f' - JSON output: {paths["json"]}')
print(f' - Markdown output: {paths["markdown"]}')
```
